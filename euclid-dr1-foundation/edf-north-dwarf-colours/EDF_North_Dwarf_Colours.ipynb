```python
from pathlib import Path
import subprocess
import sys

branch = "research/euclid-dr1-foundation-day01-20261114"
checkout = Path("daily-astro-notebooks").resolve()
if not checkout.exists():
    subprocess.run(
        ["git", "clone", "--depth", "1", "--filter=blob:none", "--sparse", "--branch", branch,
         "https://github.com/Biswajit1999/daily-astro-notebooks.git", str(checkout)],
        check=True,
    )
analysis = checkout / "euclid-dr1-foundation" / "edf-north-dwarf-colours"
subprocess.run(
    ["git", "-C", str(checkout), "sparse-checkout", "set", str(analysis.relative_to(checkout)),
     "euclid-dr1-foundation/euclid-dark-universe"],
    check=True,
)
cache = Path("edf-data").resolve()
output = cache / "results"
```

```python
for script in ("fetch_cutouts.py", "fetch_psf.py"):
    subprocess.run([sys.executable, str(analysis / script), "--cache", str(cache)], check=True)
subprocess.run(
    [sys.executable, str(analysis / "measure_colours.py"), "--cache", str(cache), "--output", str(output)],
    check=True,
)
```

```python
import json
from IPython.display import FileLink, display

rows = json.loads((output / "colour_measurements.json").read_text())
for label, row in zip(("A1", "A2", "B1", "B2", "C1", "C2"), rows):
    m = row["result"]
    print(f"{label}  {m['baseline_mag']:+.3f} mag  "
          f"PSF/sky {m['sensitivity_min_mag']:+.3f} to {m['sensitivity_max_mag']:+.3f}")
for path in sorted(output.iterdir()):
    display(FileLink(str(path.relative_to(Path.cwd()))))
```
