Measure which public DESI galaxies and random pairs remain when restricted to the Euclid footprint.
The notebook reads approximately 1.72 GB of public FITS files, checks their SHA-256 hashes, counts pairs at 95–115 h⁻¹ Mpc, and compares the Euclid masks against three compact control regions.

```python
from pathlib import Path
import subprocess
import sys

branch = "research/euclid-dr1-foundation-day01-20261114"
repo = Path("daily-astro-notebooks").resolve()
relative = Path("euclid-dr1-foundation/desi-bao-window")
if not repo.exists():
    subprocess.run(
        ["git", "clone", "--branch", branch, "--depth", "1", "--filter=blob:none", "--sparse",
         "https://github.com/Biswajit1999/daily-astro-notebooks.git", str(repo)],
        check=True,
    )
subprocess.run(["git", "-C", str(repo), "sparse-checkout", "set", str(relative)], check=True)
analysis = repo / relative
cache = Path("bao-inputs").resolve()
results = Path("bao-results").resolve()
```

```python
subprocess.run([sys.executable, str(analysis/"fetch.py"), "--cache", str(cache)], check=True)
```

```python
subprocess.run(
    [sys.executable, str(analysis/"analyse.py"), "--cache", str(cache), "--output", str(results)],
    check=True,
)
```

```python
import json

report = json.loads((results/"window_measurements.json").read_text())
for z in report["bins"]:
    result = z["galaxies"]
    keep = result["inside"]["mer_wide"]["n"]
    print(f"z = {z['z']}   MER overlap {keep:,} / {result['n']:,} galaxies")
```

```python
from IPython.display import Image, display

subprocess.run(
    [sys.executable, str(analysis/"plot_window.py"), "--cache", str(cache), "--results", str(results)],
    check=True,
)
display(Image(filename=str(results/"euclid_desi_window.png")))
```
